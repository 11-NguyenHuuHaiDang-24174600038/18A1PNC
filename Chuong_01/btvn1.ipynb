```python
class HinhChuNhat:
    def __init__(self, dai=0, rong=0):
        self.dai = dai
        self.rong = rong

    def tinh_chu_vi(self):
        return 2 * (self.dai + self.rong)

    def tinh_dien_tich(self):
        return self.dai * self.rong

    def hien_thi(self):
        print("\n=== Thông tin hình chữ nhật ===")
        print(f"Chiều dài: {self.dai}")
        print(f"Chiều rộng: {self.rong}")
        print(f"Chu vi: {self.tinh_chu_vi()}")
        print(f"Diện tích: {self.tinh_dien_tich()}")
```

```python
hcn = HinhChuNhat()
hcn.dai=10
hcn.rong=5
hcn.hien_thi()
```

```python
class ThiSinh:
    def __init__(self):
        self.ho_ten = ""
        self.toan = 0.0
        self.ly = 0.0
        self.hoa = 0.0

    def nhap_thong_tin(self):
        self.ho_ten = input("Nhập họ tên thí sinh: ")
        while True:
            try:
                self.toan = float(input("Nhập điểm Toán: "))
                self.ly = float(input("Nhập điểm Lý: "))
                self.hoa = float(input("Nhập điểm Hóa: "))
                break
            except ValueError:
                print("Vui lòng nhập số hợp lệ cho điểm!")

    def tinh_tong_diem(self):
        return self.toan + self.ly + self.hoa

    def hien_thi(self):
        print(f"Họ tên: {self.ho_ten}")
        print(f"Toán: {self.toan}, Lý: {self.ly}, Hóa: {self.hoa}")
        print(f"→ Tổng điểm: {self.tinh_tong_diem()}")
```

Danh sách thí sinh

```python
ds_thi_sinh = []

while True:
    ts = ThiSinh()
    ts.nhap_thong_tin()
    ds_thi_sinh.append(ts)

    tiep_tuc = input("Bạn có muốn nhập thêm thí sinh khác không? (y/n): ").lower()
    if tiep_tuc != 'y':
        break

print("\nĐÃ NHẬP DANH SÁCH THÍ SINH HOÀN TẤT!\n")

```

Sắp xếp và hiển thị theo thứ tự điểm giảm dần

```python
ds_thi_sinh.sort(key=lambda ts: ts.tinh_tong_diem(), reverse=True)
print("\n=== DANH SÁCH THÍ SINH (TỪ CAO XUỐNG THẤP) ===")
thu_hang = 1
for ts in ds_thi_sinh:
    print(f"\nThứ hạng {thu_hang}:")
    print(f"Họ tên: {ts.ho_ten}")
    print(f"Điểm Toán: {ts.toan}")
    print(f"Điểm Lý: {ts.ly}")
    print(f"Điểm Hóa: {ts.hoa}")
    print(f"→ Tổng điểm: {ts.tinh_tong_diem()}")
    print("-" * 40)  # In ra đường kẻ ngăn cách
    thu_hang += 1
```

Lọc danh sách trúng tuyển

```python
diem_chuan = float(input("\nNhập điểm chuẩn: "))

print(f"\n=== DANH SÁCH THÍ SINH TRÚNG TUYỂN (>= {diem_chuan}) ===")
dem = 0
for ts in ds_thi_sinh:
    if ts.tinh_tong_diem() >= diem_chuan:
        ts.hien_thi()
        print("-" * 40)
        dem += 1

if dem == 0:
    print("Không có thí sinh nào trúng tuyển.")
```

```python
class PhanSo:
    def __init__(self, tu=0, mau=1):
        self.tu = tu
        self.mau = mau

    def kiem_tra_hop_le(self):
        if self.mau == 0:
            print("Mẫu số không được bằng 0!")
            return False
        return True

    def nhap(self):
        self.tu = int(input("Nhập tử số: "))
        self.mau = int(input("Nhập mẫu số: "))
        if not self.kiem_tra_hop_le():
            print("Vui lòng nhập lại.")
            self.nhap()

    def hien_thi(self):
        if self.kiem_tra_hop_le():
            print(f"Phân số: {self.tu}/{self.mau}")
```

```python
class Stack:
    def __init__(self, size):
        self.size = size
        self.stack = [0.0] * size
        self.top = -1  

    def isEmpty(self):
        return self.top == -1

    def isFull(self):
        return self.top == self.size - 1

    def push(self, value):
        if self.isFull():
            print("không thể thêm phần tử.")
        else:
            self.top += 1
            self.stack[self.top] = value
            print(f"Đã thêm {value} vào ngăn xếp.")

    def pop(self):
        if self.isEmpty():
            print("không thể lấy phần tử.")
            return None
        else:
            value = self.stack[self.top]
            self.top -= 1
            print(f" Đã lấy {value} ra khỏi ngăn xếp.")
            return value

    def hien_thi(self):
        if self.isEmpty():
            print("Ngăn xếp hiện đang rỗng.")
        else:
            print("Các phần tử trong ngăn xếp (từ dưới lên trên):")
            for i in range(self.top + 1):
                print(f"  {self.stack[i]}")
```

```python
st = Stack(5)

st.push(10.5)
st.push(20.3)
st.push(30.1)

st.hien_thi()

st.pop()
st.hien_thi()
```

```python
class Stack:
    def __init__(self, size):
        self.size = size
        self.stack = [0.0] * size
        self.top = -1
        self.count = 0

    def isEmpty(self):
        return self.top == -1

    def isFull(self):
        return self.top == self.size - 1

    def push(self, value):
        if self.isFull():
            print("Ngăn xếp đã đầy, không thể thêm phần tử.")
        else:
            self.top += 1
            self.stack[self.top] = value
            self.count += 1
            print(f"Đã thêm {value} vào ngăn xếp.")

    def pop(self):
        if self.isEmpty():
            print("Ngăn xếp rỗng, không thể lấy phần tử.")
            return None
        else:
            value = self.stack[self.top]
            self.top -= 1
            self.count -= 1
            print(f"Đã lấy {value} ra khỏi ngăn xếp.")
            return value

    def print_stack(self):
        if self.isEmpty():
            print("Ngăn xếp rỗng.")
        else:
            print("Nội dung ngăn xếp (từ dưới lên trên):")
            for i in range(self.top + 1):
                print(f"  {self.stack[i]}")
            print(f"Tổng số phần tử: {self.count}")
```

CHẠY CHƯƠNG TRÌNH

```python
stack6 = Stack(5)
stack6.push(10.1)
stack6.push(20.2)
stack6.push(30.3)

stack6.print_stack()

stack6.pop()
stack6.print_stack()
```

```python
class Date:
    def __init__(self, ngay=1, thang=1, nam=2000):
        self.ngay = ngay
        self.thang = thang
        self.nam = nam

    def kiem_tra_hop_le(self):
        if self.thang < 1 or self.thang > 12:
            return False
        if self.ngay < 1 or self.ngay > 31:
            return False
        if self.thang in [4, 6, 9, 11] and self.ngay > 30:
            return False
        if self.thang == 2:
            if self.nam % 4 == 0 and self.nam % 100 != 0 or self.nam % 400 == 0:
                return self.ngay <= 29
            else:
                return self.ngay <= 28
        return True

    def nhap(self):
        self.ngay = int(input("Nhập ngày: "))
        self.thang = int(input("Nhập tháng: "))
        self.nam = int(input("Nhập năm: "))

        if not self.kiem_tra_hop_le():
            print(" Ngày không hợp lệ! Vui lòng nhập lại.")
            self.nhap()

    def hien_thi(self):
        print(f" Ngày: {self.ngay:02d}/{self.thang:02d}/{self.nam}")
```

```python
ngay_sinh = Date()
ngay_sinh.nhap()
ngay_sinh.hien_thi()
```

```python
import math

class TamGiac:
    def __init__(self, a=1, b=1, c=1):
        self.a = a
        self.b = b
        self.c = c

    def kiem_tra_hop_le(self):
        return (
            self.a + self.b > self.c and
            self.a + self.c > self.b and
            self.b + self.c > self.a and
            self.a > 0 and self.b > 0 and self.c > 0
        )

    def nhap(self):
        self.a = float(input("Nhập cạnh a: "))
        self.b = float(input("Nhập cạnh b: "))
        self.c = float(input("Nhập cạnh c: "))

        if not self.kiem_tra_hop_le():
            print(" Ba cạnh không hợp lệ. Vui lòng nhập lại.")
            self.nhap()


    def chu_vi(self):
        return self.a + self.b + self.c

    def dien_tich(self):
        p = self.chu_vi() / 2
        s = math.sqrt(p * (p - self.a) * (p - self.b) * (p - self.c))
        return s

    def hien_thi(self):
        if self.kiem_tra_hop_le():
            print(f"Tam giác có 3 cạnh: a={self.a}, b={self.b}, c={self.c}")
            print(f" Chu vi: {self.chu_vi():.2f}")
            print(f" Diện tích: {self.dien_tich():.2f}")
        else:
            print(" Tam giác không hợp lệ.")
```

```python
import math

class TamGiac:
    def __init__(self, a=1, b=1, c=1):
        self.a = a
        self.b = b
        self.c = c

    def kiem_tra_hop_le(self):
        return (
            self.a + self.b > self.c and
            self.a + self.c > self.b and
            self.b + self.c > self.a
        )

    def loai_tam_giac(self):
        if not self.kiem_tra_hop_le():
            return "Không hợp lệ"
        if self.a == self.b == self.c:
            return "Đều"
        elif self.a == self.b or self.a == self.c or self.b == self.c:
            return "Cân"
        elif (
            abs(self.a**2 + self.b**2 - self.c**2) < 1e-6 or
            abs(self.a**2 + self.c**2 - self.b**2) < 1e-6 or
            abs(self.b**2 + self.c**2 - self.a**2) < 1e-6
        ):
            return "Vuông"
        else:
            return "Thường"

    def chu_vi(self):
        return self.a + self.b + self.c

    def dien_tich(self):
        p = self.chu_vi() / 2
        return math.sqrt(p * (p - self.a) * (p - self.b) * (p - self.c))

    def hien_thi(self):
        if self.kiem_tra_hop_le():
            print(f" Tam giác có 3 cạnh: a={self.a}, b={self.b}, c={self.c}")
            print(f" Loại tam giác: {self.loai_tam_giac()}")
            print(f" Chu vi: {self.chu_vi():.2f}")
            print(f" Diện tích: {self.dien_tich():.2f}")
        else:
            print(" Tam giác không hợp lệ.")
```

```python
tg = TamGiac()
tg.nhap()
tg.hien_thi()
```

```python
import math

class Diem:
    def __init__(self, x=0, y=0):
        self.x = x
        self.y = y

    def nhap(self):
        self.x = float(input("Nhập hoành độ x: "))
        self.y = float(input("Nhập tung độ y: "))

    def hien_thi(self):
        print(f"({self.x}, {self.y})", end="")

    def khoang_cach(self, other):
        """Tính độ dài giữa 2 điểm"""
        dx = self.x - other.x
        dy = self.y - other.y
        return math.sqrt(dx**2 + dy**2)
```

```python
class TamGiac:
    def __init__(self, A=None, B=None, C=None):
        self.A = A if A else Diem()
        self.B = B if B else Diem()
        self.C = C if C else Diem()

    def nhap(self):
        print("Nhập tọa độ 3 đỉnh của tam giác:")
        print("Đỉnh A:")
        self.A.nhap()
        print("Đỉnh B:")
        self.B.nhap()
        print("Đỉnh C:")
        self.C.nhap()

    def chu_vi(self):
        a = self.A.khoang_cach(self.B)
        b = self.B.khoang_cach(self.C)
        c = self.C.khoang_cach(self.A)
        return a + b + c

    def dien_tich(self):
        a = self.A.khoang_cach(self.B)
        b = self.B.khoang_cach(self.C)
        c = self.C.khoang_cach(self.A)
        p = (a + b + c) / 2  # Nửa chu vi
        s = math.sqrt(p * (p - a) * (p - b) * (p - c))
        return s

    def hien_thi(self):
        print("Tam giác có các đỉnh:")
        print("A", end=""); self.A.hien_thi()
        print(", B", end=""); self.B.hien_thi()
        print(", C", end=""); self.C.hien_thi()
        print(f"\nChu vi: {self.chu_vi():.2f}")
        print(f"Diện tích: {self.dien_tich():.2f}")
```

```python
A = Diem(0, 0)
B = Diem(4, 0)
C = Diem(0, 3)
tg = TamGiac(A, B, C)

tg.hien_thi()
```
